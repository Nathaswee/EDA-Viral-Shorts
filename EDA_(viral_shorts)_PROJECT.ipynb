### **UNIVARIATE**

```python
plt.figure(figsize=(8, 6))
sns.boxplot(y=viral_shorts['retention_rate'], color='salmon', width=0.4)
plt.title('Univariate Analysis: Spread of Retention Rate', fontsize=15)
plt.ylabel('Retention Rate')
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.countplot(data=viral_shorts, x='niche', order=viral_shorts['niche'].value_counts().index, palette='viridis')
plt.title('Univariate Analysis: Number of Videos per Niche', fontsize=15)
plt.xlabel('Niche')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(viral_shorts['views_total'], kde=True, color='royalblue', bins=30)
plt.title('Univariate Analysis: Distribution of Total Views', fontsize=15)
plt.xlabel('Total Views')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure()
plt.scatter(viral_shorts['views_total'],viral_shorts['views_first_hour'])
plt.title("views_total vs views_first_hour")
plt.xlabel("views_total")
plt.ylabel("views_first_hour")
plt.show()
```
